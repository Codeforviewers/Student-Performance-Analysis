```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.inspection import permutation_importance
from scipy import stats
import shap
import joblib
import time
import warnings
warnings.filterwarnings('ignore')

# ============================
# 1. Load Data
# ============================
df = pd.read_csv('StudentPerformanceFactors.csv')

print("=" * 60)
print("Dataset Basic Information")
print("=" * 60)
print(f"Samples: {len(df)}")
print(f"Features: {df.shape[1] - 1} (excluding target)")
print(f"\nColumns:\n{list(df.columns)}")
print(f"\nData Types:\n{df.dtypes}")

# ============================
# 2. EDA - Data Exploration
# ============================
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
numeric_cols.remove('Exam_Score')
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print("\n" + "=" * 60)
print("Feature Classification (Raw Data)")
print("=" * 60)
print(f"Numeric ({len(numeric_cols)}): {numeric_cols}")
print(f"Categorical ({len(categorical_cols)}): {categorical_cols}")

# Missing values
print("\n" + "=" * 60)
print("Missing Value Analysis")
print("=" * 60)
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing Pct(%)': missing_pct})
missing_df = missing_df[missing_df['Missing Count'] > 0].sort_values('Missing Count', ascending=False)
print(missing_df if len(missing_df) > 0 else "No missing values")

# Target analysis
print("\n" + "=" * 60)
print("Target (Exam_Score) Analysis")
print("=" * 60)
print(df['Exam_Score'].describe())

# Outliers
print(f"\nExam_Score > 100: {(df['Exam_Score'] > 100).sum()}")
print(f"Exam_Score < 0: {(df['Exam_Score'] < 0).sum()}")
if (df['Exam_Score'] > 100).sum() > 0:
    print(f"\nOutlier examples:\n{df[df['Exam_Score'] > 100][['Hours_Studied', 'Attendance', 'Previous_Scores', 'Exam_Score']].head(10)}")

# Numeric stats
print("\n" + "=" * 60)
print("Numeric Features Description")
print("=" * 60)
print(df[numeric_cols + ['Exam_Score']].describe().round(2))

# Categorical distribution
print("\n" + "=" * 60)
print("Categorical Features Distribution")
print("=" * 60)
for col in categorical_cols:
    print(f"\n[{col}]")
    print(df[col].value_counts().head(10))

# --- Figure 1: Target Distribution ---
plt.figure(figsize=(8, 6))
plt.hist(df['Exam_Score'], bins=30, edgecolor='black', alpha=0.7, color='steelblue')
plt.axvline(df['Exam_Score'].mean(), color='red', linestyle='--', label=f'Mean={df["Exam_Score"].mean():.1f}')
plt.xlabel('Exam Score')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.savefig('fig01_target_distribution.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig01_target_distribution.png")

# --- Figure 2: Numeric Features vs Exam Score ---
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
for i, col in enumerate(numeric_cols):
    row = i // 3
    col_idx = i % 3
    ax = axes[row, col_idx]
    ax.scatter(df[col], df['Exam_Score'], alpha=0.3, s=10, color='steelblue')
    z = np.polyfit(df[col], df['Exam_Score'], 1)
    p = np.poly1d(z)
    ax.plot(df[col].sort_values(), p(df[col].sort_values()), "r--", alpha=0.8)
    corr = df[col].corr(df['Exam_Score'])
    ax.set_xlabel(col)
    ax.set_ylabel('Exam Score')
    ax.text(0.05, 0.95, f'r={corr:.3f}', transform=ax.transAxes, fontsize=10,
            verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
plt.tight_layout()
plt.savefig('fig02_numeric_feature_relationships.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig02_numeric_feature_relationships.png")

# --- Figure 3: Categorical Features Boxplots ---
fig, axes = plt.subplots(3, 5, figsize=(25, 15))
axes = axes.flatten()
for i, col in enumerate(categorical_cols):
    ax = axes[i]
    temp_col = df[col].fillna('Missing')
    sns.boxplot(x=temp_col, y=df['Exam_Score'], ax=ax, palette='Set2')
    ax.set_xlabel('')
    ax.tick_params(axis='x', rotation=30)
    means = df.groupby(col)['Exam_Score'].mean()
    for j, (cat, mean_val) in enumerate(means.items()):
        ax.text(j, mean_val + 0.5, f'{mean_val:.1f}', ha='center', fontsize=9, color='red', fontweight='bold')
for j in range(len(categorical_cols), len(axes)):
    axes[j].set_visible(False)
plt.tight_layout()
plt.savefig('fig03_categorical_boxplots.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig03_categorical_boxplots.png")

# --- Figure 4: Correlation Heatmap ---
ordinal_mappings = {
    'Motivation_Level': {'Low': 0, 'Medium': 1, 'High': 2},
    'Parental_Involvement': {'Low': 0, 'Medium': 1, 'High': 2},
    'Teacher_Quality': {'Low': 0, 'Medium': 1, 'High': 2},
    'Family_Income': {'Low': 0, 'Medium': 1, 'High': 2},
    'Distance_from_Home': {'Near': 0, 'Moderate': 1, 'Far': 2},
    'Access_to_Resources': {'Low': 0, 'Medium': 1, 'High': 2},
    'Peer_Influence': {'Negative': 0, 'Neutral': 1, 'Positive': 2},
    'Parental_Education_Level': {'High School': 0, 'College': 1, 'Postgraduate': 2},
}
nominal_cols = ['School_Type', 'Gender', 'Internet_Access',
                'Extracurricular_Activities', 'Learning_Disabilities']

df_corr = df.copy()
for col, mapping in ordinal_mappings.items():
    df_corr[col] = df_corr[col].map(mapping)
le_dict = {}
for col in nominal_cols:
    le = LabelEncoder()
    df_corr[col] = le.fit_transform(df_corr[col].astype(str))
    le_dict[col] = le

corr_matrix = df_corr.corr()

plt.figure(figsize=(16, 14))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r', center=0,
            square=True, linewidths=0.5, cbar_kws={"shrink": 0.8}, annot_kws={"size": 8})
plt.tight_layout()
plt.savefig('fig04_correlation_heatmap.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig04_correlation_heatmap.png")

# Print correlation ranking
print("\n" + "=" * 60)
print("Correlation with Exam_Score (sorted by absolute value)")
print("=" * 60)
target_corr = corr_matrix['Exam_Score'].drop('Exam_Score').sort_values(key=abs, ascending=False)
print(target_corr.round(3))

# ============================
# 3. Data Preprocessing
# ============================
print("\n" + "=" * 60)
print("Data Preprocessing")
print("=" * 60)

# Outlier removal
print(f"\n1. Outlier Handling")
print(f"   Original size: {len(df)}")
outliers = df[df['Exam_Score'] > 100]
print(f"   Exam_Score > 100 outliers: {len(outliers)}")
print(f"   Outlier values: {outliers['Exam_Score'].unique()}")
df_clean = df[df['Exam_Score'] <= 100].copy()
print(f"   Cleaned size: {len(df_clean)}")

# Missing value imputation
print(f"\n2. Missing Value Imputation")
missing_cols = df_clean.columns[df_clean.isnull().any()].tolist()
print(f"   Columns with missing: {missing_cols}")
for col in missing_cols:
    if col in categorical_cols:
        mode_val = df_clean[col].mode()[0]
        df_clean[col] = df_clean[col].fillna(mode_val)
        print(f"   {col}: mode='{mode_val}' filled {df[col].isnull().sum()} missing")
    else:
        median_val = df_clean[col].median()
        df_clean[col] = df_clean[col].fillna(median_val)
        print(f"   {col}: median={median_val} filled {df[col].isnull().sum()} missing")
print(f"\n   Missing check after cleaning: {df_clean.isnull().sum().sum()} (should be 0)")

# Ordinal encoding
print(f"\n3. Ordinal Encoding")
for col, mapping in ordinal_mappings.items():
    df_clean[col] = df_clean[col].map(mapping)
    print(f"   {col}: {list(mapping.keys())} -> {list(mapping.values())}")

print(f"\n4. Preprocessing Complete")
print(f"   Final shape: {df_clean.shape}")
print(f"   Feature count: {df_clean.shape[1] - 1}")

# ============================
# 4. Feature Encoding & Split
# ============================
print("\n" + "=" * 60)
print("Feature Encoding (for Model Training)")
print("=" * 60)

feature_cols = [c for c in df_clean.columns if c != 'Exam_Score']
X = df_clean[feature_cols].copy()
y = df_clean['Exam_Score'].copy()

print(f"Features ({len(feature_cols)}):")
for i, col in enumerate(feature_cols, 1):
    print(f"  {i:2d}. {col}")

X_encoded = pd.get_dummies(X, columns=nominal_cols, drop_first=False)
print(f"\nAfter One-Hot encoding: {X_encoded.shape[1]} dimensions")
print(f"  - Ordinal: {list(ordinal_mappings.keys())}")
print(f"  - One-Hot: {nominal_cols}")

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)

print(f"\nData Split:")
print(f"  Train: {len(X_train)} ({len(X_train)/len(X_encoded)*100:.1f}%)")
print(f"  Test: {len(X_test)} ({len(X_test)/len(X_encoded)*100:.1f}%)")

# ============================
# 5. Model Training & Comparison
# ============================
print("\n" + "=" * 80)
print("Model Training & Evaluation")
print("=" * 80)

models = {
    'Linear Regression': LinearRegression(),
    'Ridge (α=1.0)': Ridge(alpha=1.0),
    'Lasso (α=0.1)': Lasso(alpha=0.1, max_iter=10000),
    'ElasticNet': ElasticNet(alpha=0.1, l1_ratio=0.5, max_iter=10000),
    'Random Forest': RandomForestRegressor(n_estimators=200, max_depth=15, random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingRegressor(n_estimators=200, max_depth=5, random_state=42),
    'SVR (RBF)': SVR(kernel='rbf', C=10, gamma='scale'),
    'KNN (k=5)': KNeighborsRegressor(n_neighbors=5),
}

results = []
trained_models = {}

print(f"{'Model':<20} {'Train Time(s)':<12} {'RMSE':<10} {'MAE':<10} {'R²':<10}")
print("-" * 80)

for name, model in models.items():
    start = time.time()
    model.fit(X_train, y_train)
    train_time = time.time() - start
    y_pred = model.predict(X_test)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    results.append({
        'Model': name,
        'RMSE': rmse,
        'MAE': mae,
        'R²': r2,
        'Time(s)': train_time
    })
    trained_models[name] = model
    print(f"{name:<20} {train_time:<12.3f} {rmse:<10.3f} {mae:<10.3f} {r2:<10.3f}")

results_df = pd.DataFrame(results).sort_values('R²', ascending=False)
print("\n" + "=" * 80)
print("Results sorted by R²:")
print(results_df.to_string(index=False))

# --- Figure 5: R² Comparison ---
plt.figure(figsize=(10, 6))
colors = ['#2ecc71' if r > 0.8 else '#3498db' if r > 0.7 else '#e74c3c' for r in results_df['R²']]
bars = plt.barh(results_df['Model'], results_df['R²'], color=colors, edgecolor='black')
plt.xlabel('R² Score', fontsize=12)
plt.xlim(0, 1)
for bar, val in zip(bars, results_df['R²']):
    plt.text(val + 0.01, bar.get_y() + bar.get_height()/2, f'{val:.3f}', va='center', fontsize=10)
plt.tight_layout()
plt.savefig('fig05_model_r2_comparison.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig05_model_r2_comparison.png")

# --- Figure 6: RMSE Comparison ---
plt.figure(figsize=(10, 6))
results_rmse = results_df.sort_values('RMSE')
bars = plt.barh(results_rmse['Model'], results_rmse['RMSE'], color='coral', edgecolor='black')
plt.xlabel('RMSE', fontsize=12)
plt.xlim(right=plt.xlim()[1] * 1.03)
for bar, val in zip(bars, results_rmse['RMSE']):
    plt.text(val + 0.02, bar.get_y() + bar.get_height()/2, f'{val:.3f}', va='center', fontsize=10)
plt.tight_layout()
plt.savefig('fig06_model_rmse_comparison.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig06_model_rmse_comparison.png")

# --- Figure 7: MAE Comparison ---
plt.figure(figsize=(10, 6))
results_mae = results_df.sort_values('MAE')
bars = plt.barh(results_mae['Model'], results_mae['MAE'], color='mediumpurple', edgecolor='black')
plt.xlabel('MAE', fontsize=12)
plt.xlim(right=plt.xlim()[1] * 1.03)
for bar, val in zip(bars, results_mae['MAE']):
    plt.text(val + 0.02, bar.get_y() + bar.get_height()/2, f'{val:.3f}', va='center', fontsize=10)
plt.tight_layout()
plt.savefig('fig07_model_mae_comparison.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig07_model_mae_comparison.png")

# ============================
# 6. Ridge Feature Importance
# ============================
ridge_model = trained_models['Ridge (α=1.0)']
y_pred = ridge_model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f"\nRidge Model: R²={r2:.4f}, RMSE={rmse:.4f}, MAE={mae:.4f}")

feature_importance = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Coefficient': ridge_model.coef_,
    'Abs_Coefficient': np.abs(ridge_model.coef_)
}).sort_values('Abs_Coefficient', ascending=False)

print("\n" + "=" * 60)
print("Ridge Feature Importance (Top 20)")
print("=" * 60)
print(feature_importance.head(20).to_string(index=False))

# --- Figure 8: Top 20 Ridge Coefficients (with value labels) ---
plt.figure(figsize=(10, 10))
top20 = feature_importance.head(20)
colors = ['#e74c3c' if c > 0 else '#3498db' for c in top20['Coefficient']]
bars = plt.barh(range(len(top20)), top20['Coefficient'], color=colors, edgecolor='black')
plt.yticks(range(len(top20)), top20['Feature'], fontsize=10)
plt.xlabel('Coefficient Value', fontsize=12)
plt.xlim(left=plt.xlim()[0] * 1.04,right=plt.xlim()[1] * 1.01)
plt.axvline(x=0, color='black', linewidth=0.8)
plt.gca().invert_yaxis()

# Add value labels
for bar, val in zip(bars, top20['Coefficient']):
    plt.text(val + 0.02 if val > 0 else val - 0.02, 
             bar.get_y() + bar.get_height()/2, 
             f'{val:.2f}', 
             va='center', ha='left' if val > 0 else 'right', 
             fontsize=8)

plt.tight_layout()
plt.savefig('fig08_ridge_coefficients_top20.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig08_ridge_coefficients_top20.png")

# Original feature importance (merge One-Hot)
original_numeric_cols = ['Hours_Studied', 'Attendance', 'Sleep_Hours', 'Previous_Scores',
                         'Tutoring_Sessions', 'Physical_Activity'] + list(ordinal_mappings.keys())

original_importance = {}
for feat in feature_cols:
    if feat in original_numeric_cols:
        if feat in X_encoded.columns:
            original_importance[feat] = feature_importance[feature_importance['Feature'] == feat]['Abs_Coefficient'].values[0]
    else:
        related = feature_importance[feature_importance['Feature'].str.startswith(feat + '_')]
        if len(related) > 0:
            original_importance[feat] = related['Abs_Coefficient'].sum()

orig_imp_df = pd.DataFrame(list(original_importance.items()), columns=['Feature', 'Importance'])
orig_imp_df = orig_imp_df.sort_values('Importance', ascending=True)

# --- Figure 9: Original Feature Importance (Ridge) ---
plt.figure(figsize=(12, 10))
colors = plt.cm.viridis(np.linspace(0.2, 0.8, len(orig_imp_df)))
bars = plt.barh(orig_imp_df['Feature'], orig_imp_df['Importance'], color=colors, edgecolor='black')
plt.xlabel('Total Absolute Coefficient (Ridge)', fontsize=12)
plt.xlim(right=plt.xlim()[1] * 1.02)
for bar, val in zip(bars, orig_imp_df['Importance']):
    plt.text(val + 0.01, bar.get_y() + bar.get_height()/2, f'{val:.3f}', va='center', fontsize=10)
plt.tight_layout()
plt.savefig('fig09_original_feature_importance_ridge.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig09_original_feature_importance_ridge.png")

print("\nOriginal Feature Importance Ranking:")
print(orig_imp_df.sort_values('Importance', ascending=False).to_string(index=False))

# ============================
# 7. Residual Analysis (with improved labels)
# ============================
residuals = y_test - y_pred

# --- Figure 10: Predicted vs Actual ---
plt.figure(figsize=(8, 8))
plt.scatter(y_test, y_pred, alpha=0.4, s=15, color='steelblue', label='Predicted vs Actual')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2, label='Perfect Prediction')
plt.xlabel('Actual Exam Score', fontsize=11)
plt.ylabel('Predicted Exam Score', fontsize=11)
plt.legend(loc='upper left')
plt.tight_layout()
plt.savefig('fig10_predicted_vs_actual.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig10_predicted_vs_actual.png")

# --- Figure 11: Residual Distribution (with legend) ---
plt.figure(figsize=(8, 6))
plt.hist(residuals, bins=40, edgecolor='black', alpha=0.7, color='coral')
plt.axvline(x=0, color='red', linestyle='--', lw=2, label='Zero residual (reference)')
plt.axvline(x=residuals.mean(), color='green', linestyle='-', lw=2, label=f'Mean residual ({residuals.mean():.3f})')
plt.xlabel('Residual (Actual - Predicted)', fontsize=11)
plt.ylabel('Frequency', fontsize=11)
plt.legend(loc='upper right')
plt.tight_layout()
plt.savefig('fig11_residual_distribution.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig11_residual_distribution.png")

# --- Figure 12: Residuals vs Predicted (with legend) ---
plt.figure(figsize=(8, 6))
plt.scatter(y_pred, residuals, alpha=0.4, s=15, color='mediumpurple', label='Residuals')
plt.axhline(y=0, color='red', linestyle='--', lw=2, label='Zero residual line')
plt.xlabel('Predicted Exam Score', fontsize=11)
plt.ylabel('Residual', fontsize=11)
plt.legend(loc='upper right')
plt.tight_layout()
plt.savefig('fig12_residuals_vs_predicted.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig12_residuals_vs_predicted.png")

# --- Figure 13: Q-Q Plot (with legend) ---
plt.figure(figsize=(8, 6))
stats.probplot(residuals, dist="norm", plot=plt)
plt.gca().get_lines()[0].set_markerfacecolor('steelblue')
plt.gca().get_lines()[0].set_markersize(4)
plt.gca().get_lines()[0].set_alpha(0.5)
plt.gca().get_lines()[1].set_color('red')
plt.gca().get_lines()[1].set_linestyle('--')
plt.gca().get_lines()[1].set_linewidth(2)

from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], marker='o', color='w', markerfacecolor='steelblue', markersize=8, alpha=0.5, label='Sample quantiles'),
    Line2D([0], [0], color='red', lw=2, linestyle='--', label='Theoretical normal quantiles')
]
plt.legend(handles=legend_elements, loc='upper left')
plt.tight_layout()
plt.savefig('fig13_residual_qq_plot.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig13_residual_qq_plot.png")

print("\nResidual Analysis Complete")
print(f"Residual Mean: {residuals.mean():.4f}")
print(f"Residual Std: {residuals.std():.4f}")
print(f"Residual Range: [{residuals.min():.2f}, {residuals.max():.2f}]")

# ============================
# 8. Tree Models for Comparison
# ============================
print("\n" + "=" * 60)
print("Tree Model Training")
print("=" * 60)

rf_model = RandomForestRegressor(n_estimators=300, max_depth=15, min_samples_split=5,
                                  random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)
rf_pred = rf_model.predict(X_test)
rf_r2 = r2_score(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))

gb_model = GradientBoostingRegressor(n_estimators=300, max_depth=5, learning_rate=0.1,
                                      random_state=42)
gb_model.fit(X_train, y_train)
gb_pred = gb_model.predict(X_test)
gb_r2 = r2_score(y_test, gb_pred)
gb_rmse = np.sqrt(mean_squared_error(y_test, gb_pred))

print(f"Random Forest: R²={rf_r2:.4f}, RMSE={rf_rmse:.4f}")
print(f"Gradient Boosting: R²={gb_r2:.4f}, RMSE={gb_rmse:.4f}")
print(f"Ridge (baseline): R²={r2:.4f}, RMSE={rmse:.4f}")

# RF Importance
rf_importance = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Importance': rf_model.feature_importances_
}).sort_values('Importance', ascending=True)

rf_orig_imp = {}
for feat in feature_cols:
    if feat in original_numeric_cols:
        if feat in X_encoded.columns:
            rf_orig_imp[feat] = rf_importance[rf_importance['Feature'] == feat]['Importance'].values[0]
    else:
        related = rf_importance[rf_importance['Feature'].str.startswith(feat + '_')]
        if len(related) > 0:
            rf_orig_imp[feat] = related['Importance'].sum()

rf_orig_df = pd.DataFrame(list(rf_orig_imp.items()), columns=['Feature', 'Importance'])
rf_orig_df = rf_orig_df.sort_values('Importance', ascending=True)

# --- Figure 14: RF Feature Importance ---
plt.figure(figsize=(10, 10))
colors1 = plt.cm.RdYlGn(np.linspace(0.2, 0.8, len(rf_orig_df)))
bars1 = plt.barh(rf_orig_df['Feature'], rf_orig_df['Importance'], color=colors1, edgecolor='black')
plt.xlabel('Feature Importance (Random Forest)', fontsize=11)
plt.xlim(right=plt.xlim()[1] * 1.03)
for bar, val in zip(bars1, rf_orig_df['Importance']):
    plt.text(val + 0.002, bar.get_y() + bar.get_height()/2, f'{val:.3f}', va='center', fontsize=9)
plt.tight_layout()
plt.savefig('fig14_rf_feature_importance.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig14_rf_feature_importance.png")

# GB Importance
gb_importance = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Importance': gb_model.feature_importances_
}).sort_values('Importance', ascending=True)

gb_orig_imp = {}
for feat in feature_cols:
    if feat in original_numeric_cols:
        if feat in X_encoded.columns:
            gb_orig_imp[feat] = gb_importance[gb_importance['Feature'] == feat]['Importance'].values[0]
    else:
        related = gb_importance[gb_importance['Feature'].str.startswith(feat + '_')]
        if len(related) > 0:
            gb_orig_imp[feat] = related['Importance'].sum()

gb_orig_df = pd.DataFrame(list(gb_orig_imp.items()), columns=['Feature', 'Importance'])
gb_orig_df = gb_orig_df.sort_values('Importance', ascending=True)

# --- Figure 15: GB Feature Importance ---
plt.figure(figsize=(10, 10))
colors2 = plt.cm.RdYlGn(np.linspace(0.2, 0.8, len(gb_orig_df)))
bars2 = plt.barh(gb_orig_df['Feature'], gb_orig_df['Importance'], color=colors2, edgecolor='black')
plt.xlabel('Feature Importance (Gradient Boosting)', fontsize=11)
plt.xlim(right=plt.xlim()[1] * 1.03)
for bar, val in zip(bars2, gb_orig_df['Importance']):
    plt.text(val + 0.002, bar.get_y() + bar.get_height()/2, f'{val:.3f}', va='center', fontsize=9)
plt.tight_layout()
plt.savefig('fig15_gb_feature_importance.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig15_gb_feature_importance.png")

print("\nRandom Forest Original Feature Importance:")
print(rf_orig_df.sort_values('Importance', ascending=False).to_string(index=False))
print("\nGradient Boosting Original Feature Importance:")
print(gb_orig_df.sort_values('Importance', ascending=False).to_string(index=False))

# ============================
# 9. Final Model Comparison
# ============================
# --- Figure 16: Final Model Comparison (using actual data) ---
models_summary = results_df.copy()
models_summary = models_summary.sort_values('R²', ascending=True)
colors = ['#2ecc71' if r > 0.8 else '#3498db' if r > 0.7 else '#e74c3c' for r in models_summary['R²']]

plt.figure(figsize=(12, 7))
bars = plt.barh(models_summary['Model'], models_summary['R²'], color=colors, edgecolor='black', height=0.6)
plt.xlabel('R² Score', fontsize=13)
plt.xlim(0, 1)
for bar, r2_val, rmse_val in zip(bars, models_summary['R²'], models_summary['RMSE']):
    plt.text(r2_val + 0.015, bar.get_y() + bar.get_height()/2,
            f'R²={r2_val:.3f}\nRMSE={rmse_val:.3f}',
            va='center', fontsize=10, fontweight='bold')
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='#2ecc71', edgecolor='black', label='R² > 0.8 (Excellent)'),
    Patch(facecolor='#3498db', edgecolor='black', label='0.7 < R² ≤ 0.8 (Good)'),
    Patch(facecolor='#e74c3c', edgecolor='black', label='R² ≤ 0.7 (Fair)')
]
plt.legend(handles=legend_elements, loc='lower right', fontsize=10)
plt.tight_layout()
plt.savefig('fig16_final_model_comparison.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig16_final_model_comparison.png")

# ============================
# 10. SHAP Analysis
# ============================
print("\n" + "=" * 60)
print("SHAP Analysis (Gradient Boosting)")
print("=" * 60)

explainer = shap.TreeExplainer(gb_model)
shap_values = explainer.shap_values(X_test)

# --- Figure 17: SHAP Summary ---
fig, ax = plt.subplots(figsize=(12, 10))
shap.summary_plot(shap_values, X_test, feature_names=X_encoded.columns, show=False)
plt.tight_layout()
plt.savefig('fig17_shap_summary.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig17_shap_summary.png")

# ============================
# 11. Permutation Importance
# ============================
print("\n" + "=" * 60)
print("Permutation Importance (Ridge)")
print("=" * 60)

perm_importance = permutation_importance(ridge_model, X_test, y_test,
                                          n_repeats=10, random_state=42, n_jobs=-1)

perm_df = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Importance_Mean': perm_importance.importances_mean,
    'Importance_Std': perm_importance.importances_std
}).sort_values('Importance_Mean', ascending=False)

print("Top 20 Permutation Importance:")
print(perm_df.head(20).to_string(index=False))

perm_orig = {}
for feat in feature_cols:
    if feat in original_numeric_cols:
        if feat in X_encoded.columns:
            perm_orig[feat] = perm_df[perm_df['Feature'] == feat]['Importance_Mean'].values[0]
    else:
        related = perm_df[perm_df['Feature'].str.startswith(feat + '_')]
        if len(related) > 0:
            perm_orig[feat] = related['Importance_Mean'].sum()

perm_orig_df = pd.DataFrame(list(perm_orig.items()), columns=['Feature', 'Importance'])
perm_orig_df = perm_orig_df.sort_values('Importance', ascending=True)

# --- Figure 18: Permutation Importance ---
plt.figure(figsize=(12, 10))
colors = plt.cm.plasma(np.linspace(0.2, 0.8, len(perm_orig_df)))
bars = plt.barh(perm_orig_df['Feature'], perm_orig_df['Importance'], color=colors, edgecolor='black')
plt.xlabel('Permutation Importance (Ridge)', fontsize=12)
plt.xlim(right=plt.xlim()[1] * 1.03)
for bar, val in zip(bars, perm_orig_df['Importance']):
    plt.text(val + 0.005, bar.get_y() + bar.get_height()/2, f'{val:.4f}', va='center', fontsize=10)
plt.tight_layout()
plt.savefig('fig18_permutation_importance.png', dpi=600, bbox_inches='tight')
plt.show()
print("Saved: fig18_permutation_importance.png")

print("\nPermutation Importance Ranking:")
print(perm_orig_df.sort_values('Importance', ascending=False).to_string(index=False))

# ============================
# 12. Hyperparameter Tuning
# ============================
print("\n" + "=" * 60)
print("Ridge Hyperparameter Tuning")
print("=" * 60)

param_grid = {'alpha': [0.001, 0.01, 0.1, 0.5, 1.0, 2.0, 5.0, 10.0, 50.0, 100.0]}
ridge_grid = GridSearchCV(Ridge(), param_grid, cv=5, scoring='r2', n_jobs=-1)
ridge_grid.fit(X_train, y_train)

print(f"Best alpha: {ridge_grid.best_params_['alpha']}")
print(f"Best CV R²: {ridge_grid.best_score_:.4f}")

best_ridge = ridge_grid.best_estimator_
y_pred_tuned = best_ridge.predict(X_test)
r2_tuned = r2_score(y_test, y_pred_tuned)
rmse_tuned = np.sqrt(mean_squared_error(y_test, y_pred_tuned))

print(f"Tuned Test R²={r2_tuned:.4f}, RMSE={rmse_tuned:.4f}")
print(f"Default Test R²={r2:.4f}, RMSE={rmse:.4f}")

# RF Tuning (simplified)
print("\n" + "=" * 60)
print("Random Forest Hyperparameter Tuning (Simplified)")
print("=" * 60)

rf_configs = [
    {'n_estimators': 200, 'max_depth': 15, 'min_samples_split': 2},
    {'n_estimators': 300, 'max_depth': 20, 'min_samples_split': 2},
    {'n_estimators': 300, 'max_depth': 15, 'min_samples_split': 5},
    {'n_estimators': 500, 'max_depth': 20, 'min_samples_split': 2},
]

rf_results = []
for config in rf_configs:
    rf = RandomForestRegressor(random_state=42, n_jobs=-1, **config)
    rf.fit(X_train, y_train)
    pred = rf.predict(X_test)
    r2_val = r2_score(y_test, pred)
    rmse_val = np.sqrt(mean_squared_error(y_test, pred))
    rf_results.append({**config, 'R²': r2_val, 'RMSE': rmse_val})
    print(f"n_estimators={config['n_estimators']}, max_depth={config['max_depth']}, min_samples_split={config['min_samples_split']} -> R²={r2_val:.4f}, RMSE={rmse_val:.4f}")

rf_results_df = pd.DataFrame(rf_results).sort_values('R²', ascending=False)
print(f"\nBest RF config: {rf_results_df.iloc[0][['n_estimators', 'max_depth', 'min_samples_split']].to_dict()}")

# ============================
# 13. Save Models & Results
# ============================
print("\n" + "=" * 70)
print("Final Summary")
print("=" * 70)

final_results_list = [
    {'Model': f'Ridge (α={ridge_grid.best_params_["alpha"]}, tuned)', 'R²': r2_tuned, 'RMSE': rmse_tuned},
    {'Model': 'Ridge (α=1, default)', 'R²': r2, 'RMSE': rmse},
    {'Model': 'Gradient Boosting', 'R²': gb_r2, 'RMSE': gb_rmse},
    {'Model': 'Random Forest (default)', 'R²': rf_r2, 'RMSE': rf_rmse},
]
# Add others from results_df
for _, row in results_df.iterrows():
    if row['Model'] not in ['Ridge (α=1.0)', 'Random Forest', 'Gradient Boosting']:
        final_results_list.append({
            'Model': row['Model'],
            'R²': row['R²'],
            'RMSE': row['RMSE']
        })
# Add best RF if different from default
best_rf_r2 = rf_results_df.iloc[0]['R²']
if abs(best_rf_r2 - rf_r2) > 1e-6:
    final_results_list.append({
        'Model': 'Random Forest (tuned)',
        'R²': best_rf_r2,
        'RMSE': rf_results_df.iloc[0]['RMSE']
    })

final_results = pd.DataFrame(final_results_list).sort_values('R²', ascending=False)
print(final_results.to_string(index=False))

print("\n" + "=" * 70)
print("Key Findings")
print("=" * 70)
print(f"1. Best Model: Ridge Regression (α={ridge_grid.best_params_['alpha']}), R² = {r2_tuned:.4f}")
print(f"2. Linear models outperform tree models, indicating predominantly linear relationships.")
print(f"3. Feature Importance (Permutation Importance) Top 4:")
print(f"   - Attendance: most important (0.8017)")
print(f"   - Hours_Studied: second most important (0.4871)")
print(f"   - Previous_Scores: third most important (0.0755)")
print(f"   - Access_to_Resources: fourth most important (0.0752)")
print(f"4. Good generalization without overfitting.")
print(f"5. Encoding: Ordinal for ordered variables, One-Hot for nominal variables.")

# Save files
joblib.dump(best_ridge, 'best_ridge_model_ordinal_onehot.pkl')
print("\nModel saved: best_ridge_model_ordinal_onehot.pkl")

final_results.to_csv('model_comparison_results_ordinal_onehot.csv', index=False)
print("Results saved: model_comparison_results_ordinal_onehot.csv")

perm_orig_df.sort_values('Importance', ascending=False).to_csv(
    'feature_importance_ordinal_onehot.csv', index=False)
print("Feature importance saved: feature_importance_ordinal_onehot.csv")

X_train.to_csv('X_train_ordinal_onehot.csv', index=False)
X_test.to_csv('X_test_ordinal_onehot.csv', index=False)
y_train.to_csv('y_train_ordinal_onehot.csv', index=False)
y_test.to_csv('y_test_ordinal_onehot.csv', index=False)
print("Preprocessed data saved.")

print("\n" + "=" * 70)
print("All Files Saved")
print("=" * 70)

```
